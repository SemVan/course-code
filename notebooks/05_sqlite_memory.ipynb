# 05. Память чат-бота в SQLite

Изолированная история пользователей с ограничением длины.

Стек: Python standard library. Выполняйте ячейки сверху вниз.

```python
import sqlite3, time
con=sqlite3.connect(':memory:')
con.execute('CREATE TABLE messages(user_id TEXT, role TEXT, content TEXT, created_at INTEGER)')

def remember(user_id, role, content):
    con.execute('INSERT INTO messages VALUES(?,?,?,?)',(user_id,role,content,int(time.time())))
    con.commit()

def recent(user_id, limit=6):
    rows=con.execute('SELECT role,content FROM messages WHERE user_id=? ORDER BY created_at DESC LIMIT ?',(user_id,limit)).fetchall()
    return [dict(role=r,content=c) for r,c in reversed(rows)]

remember('u1','user','Нужен краткий формат')
remember('u1','assistant','Принято')
recent('u1')
```
