# 07. Симулятор Telegram update

Разбираем события без токена, сети и реального бота.

Стек: Python standard library. Выполняйте ячейки сверху вниз.

```python
updates=[
 {'update_id':1,'message':{'chat':{'id':101},'text':'/help'}},
 {'update_id':2,'message':{'chat':{'id':101},'text':'Как оформить отпуск?'}},
 {'update_id':3,'edited_message':{'chat':{'id':101},'text':'исправлено'}},
]
def normalize(update):
    msg=update.get('message')
    if not isinstance(msg,dict): return None
    chat=msg.get('chat') or {}; text=msg.get('text')
    if not isinstance(chat.get('id'),int) or not isinstance(text,str): return None
    return {'update_id':update.get('update_id'),'chat_id':chat['id'],'text':text[:1000]}

[normalize(u) for u in updates]
```
