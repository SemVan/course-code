# 09. Ограниченный агент как конечный автомат

Детерминированная оболочка вокруг решения модели.

Стек: Python standard library. Выполняйте ячейки сверху вниз.

```python
from enum import Enum
class State(Enum): PLAN='plan'; ACT='act'; VERIFY='verify'; DONE='done'; ESCALATE='escalate'
ALLOWED={'search_local','draft'}

def agent(plan,max_steps=4):
    state=State.PLAN; log=[]
    for step in plan[:max_steps]:
        tool=step.get('tool')
        if tool not in ALLOWED:
            return State.ESCALATE, log+[f'blocked:{tool}']
        state=State.ACT; log.append(f'act:{tool}')
        state=State.VERIFY; log.append('verified')
    if len(plan)>max_steps: return State.ESCALATE,log+['step_budget_exceeded']
    return State.DONE,log

agent([{'tool':'search_local'},{'tool':'draft'}])
```
