# 02. Локальная LLM через Ollama

HTTP-вызов без SDK. Если Ollama не запущен, ячейка выводит понятную диагностику.

Стек: Python standard library. Выполняйте ячейки сверху вниз.

```python
import json, urllib.request, urllib.error, time

OLLAMA_URL = 'http://localhost:11434/api/chat'
MODEL = 'qwen2.5:3b'  # замените на уже установленную открытую instruct-модель

def chat(prompt, timeout=60):
    body = json.dumps({'model': MODEL, 'messages': [{'role':'user','content':prompt}], 'stream': False}).encode()
    req = urllib.request.Request(OLLAMA_URL, data=body, headers={'Content-Type':'application/json'})
    started=time.perf_counter()
    try:
        with urllib.request.urlopen(req, timeout=timeout) as r:
            data=json.load(r)
        return {'text': data['message']['content'], 'seconds': round(time.perf_counter()-started,2)}
    except urllib.error.URLError as e:
        return {'error': 'Ollama недоступен. Запустите локальный сервер и проверьте имя модели.', 'detail': str(e.reason)}

chat('Верни три критерия хорошего делового письма, кратко.')
```
