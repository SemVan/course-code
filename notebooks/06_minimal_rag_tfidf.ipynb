# 06. Мини-RAG на TF-IDF

Поиск по малому корпусу без векторной БД и внешних библиотек.

Стек: Python standard library. Выполняйте ячейки сверху вниз.

```python
import math,re
from collections import Counter

docs={
 'vacation':'Отпуск согласуется с руководителем. Заявление подают не позднее чем за две недели.',
 'remote':'Удалённая работа оформляется соглашением и согласуется с руководителем.',
 'security':'Пароли нельзя отправлять в чат. Для доступа используется менеджер секретов.'}

def tokens(s): return re.findall(r'[а-яёa-z0-9]+',s.lower())
N=len(docs); df=Counter()
for text in docs.values(): df.update(set(tokens(text)))
idf={t:math.log((N+1)/(n+1))+1 for t,n in df.items()}
def vec(text):
    tf=Counter(tokens(text)); return {t:n*idf.get(t,0) for t,n in tf.items()}
def cosine(a,b):
    dot=sum(v*b.get(k,0) for k,v in a.items()); na=math.sqrt(sum(v*v for v in a.values())); nb=math.sqrt(sum(v*v for v in b.values()))
    return dot/(na*nb) if na and nb else 0
def search(q,k=2):
    qv=vec(q); scored=[(cosine(qv,vec(text)),name,text) for name,text in docs.items()]
    return sorted(scored,reverse=True)[:k]
search('Когда подавать заявление на отпуск?')
```
