# 04. Безопасный парсинг JSON

Извлекаем объект и проверяем схему без pydantic.

Стек: Python standard library. Выполняйте ячейки сверху вниз.

```python
import json
ALLOWED={'support','hr','other'}

def parse_ticket(raw):
    obj=json.loads(raw)
    if not isinstance(obj,dict): raise ValueError('Нужен JSON-объект')
    if set(obj) != {'category','priority','summary'}: raise ValueError('Неверный набор полей')
    if obj['category'] not in ALLOWED: raise ValueError('Неизвестная категория')
    if not isinstance(obj['priority'],int) or not 1 <= obj['priority'] <= 5: raise ValueError('priority: целое 1..5')
    if not isinstance(obj['summary'],str) or not 1 <= len(obj['summary']) <= 200: raise ValueError('Некорректный summary')
    return obj

parse_ticket('{"category":"support","priority":3,"summary":"Не открывается отчёт"}')
```
